```sql
SELECT  motivo_feedback,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
GROUP BY ALL
SORT BY qtde DESC
```

```sql
SELECT  'PhoneService' AS servico,
        LOWER(PhoneService) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
GROUP BY ALL
UNION ALL
SELECT  'InternetService' AS servico,
        LOWER(InternetService) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
GROUP BY ALL
```

```sql
SELECT  LOWER(MultipleLines) AS MultipleLines,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(PhoneService) = 'yes'
GROUP BY ALL
```

```sql
SELECT  'OnlineSecurity' AS servico,
        LOWER(OnlineSecurity) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(InternetService) <> 'no'
GROUP BY ALL
UNION ALL
SELECT  'OnlineBackup' AS servico,
        LOWER(OnlineBackup) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(InternetService) <> 'no'
GROUP BY ALL
UNION ALL
SELECT  'DeviceProtection' AS servico,
        LOWER(DeviceProtection) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(InternetService) <> 'no'
GROUP BY ALL
UNION ALL
SELECT  'TechSupport' AS servico,
        LOWER(TechSupport) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(InternetService) <> 'no'
GROUP BY ALL
UNION ALL
SELECT  'StreamingTV' AS servico,
        LOWER(StreamingTV) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(InternetService) <> 'no'
GROUP BY ALL
UNION ALL
SELECT  'StreamingMovies' AS servico,
        LOWER(StreamingMovies) AS contratado,
        COUNT(*) AS qtde
FROM    workspace.default.train_classificado
WHERE   LOWER(Churn) = 'yes'
        AND motivo_feedback = 'qualidade_servico'
        AND LOWER(InternetService) <> 'no'
GROUP BY ALL
```
